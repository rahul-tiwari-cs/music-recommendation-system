```python
def recommend_song(song_name, no_of_recommendations = 5, data_frame = data, feature_matrix = music_data):
    matches = data[data['name'].str.lower() == (song_name.lower().strip())]

    if matches.empty:
        return f"Error: '{song_name}' not found in the dataset."
        
    idx = matches.index[0]

    target_vector = music_data[idx:idx+1]
    similarity_scores = cosine_similarity(target_vector, music_feat).flatten()

    similarity_scores[idx] = -1

    top_indices = np.argpartition(similarity_scores, -(no_of_recommendations))[-no_of_recommendations:]
    top_indices = top_indices[np.argsort(similarity_scores[top_indices])[::-1]]

    recommendations = data.iloc[top_indices][['name', 'artists', 'year']]

    return recommendations
```
